```python
import glob
import os
import xarray as xr
import numpy as np
import pandas as pd

files = glob.glob("../dataset/*.nc")

print("Number of NetCDF files:", len(files))

for file in files:
    print(os.path.basename(file))
```

```python
all_features = []

for file in files:
    ds_temp = xr.open_dataset(file)

    features = {}

    for variable in ds_temp.data_vars:
        if variable.startswith("ssmis"):
            data = ds_temp[variable].values

            features[variable + "_mean"] = np.mean(data)
            features[variable + "_std"] = np.std(data)
            features[variable + "_min"] = np.min(data)
            features[variable + "_max"] = np.max(data)

    # Add the filename so we know which observation this row came from
    features["filename"] = os.path.basename(file)

    all_features.append(features)

    ds_temp.close()

feature_df = pd.DataFrame(all_features)

print("Dataset shape:", feature_df.shape)

feature_df.head()
```

```python
print("Number of rows:", len(feature_df))
print("Number of columns:", len(feature_df.columns))

print("\nMissing values:")
print(feature_df.isnull().sum().sum())
```

```python
feature_df.to_csv("../dataset/ssmis_features.csv", index=False)

print("SSMIS feature dataset saved successfully!")
```
